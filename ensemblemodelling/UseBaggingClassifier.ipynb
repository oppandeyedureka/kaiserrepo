```python
#import libraries and resoruces
import numpy as np
import pandas as pd
from sklearn.ensemble import BaggingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.metrics import f1_score, r2_score, accuracy_score,classification_report
from sklearn import datasets
from sklearn.model_selection import train_test_split

#load dataset
cancerds = datasets.load_breast_cancer()

#load features
X = cancerds.data
Y = cancerds.target


#Train test split
X_Train, X_Test, Y_Train, Y_Test = train_test_split(X,Y, random_state=42, test_size=0.30)

#Chose Base Classifier and create Ensemble model

# base_classifier = DecisionTreeClassifier(max_depth=5, max_leaf_nodes=10)
# base_classifier = svmmodel = SVC(kernel='linear')
base_classifier = LogisticRegression(max_iter=18000)
en_model = BaggingClassifier(base_classifier, n_estimators=10, random_state=42)
en_model.fit(X_Train, Y_Train)

#Predictions and Metrics
predictions = en_model.predict(X_Test)
accuracy = accuracy_score(Y_Test, predictions)
print("Accuracy:", accuracy)
print(classification_report(Y_Test, predictions))
```

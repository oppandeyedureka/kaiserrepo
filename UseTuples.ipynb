Create examples of using Tuples

```python
#create a Tuple
tuple01 = () #empty Tuple
tuple02 = ("Hello","Welcome","Hi")#values in tuple
print(tuple02)

#tuple02.insert(1,"Bye")#no Insert operation - Immutable
#tuple02.append("Bye")#no Insert operation - Immutable
#tuple02[2] = "Bye"#No Update Operation

print(tuple02[0])
print(tuple02[-1])
print(type(tuple02))

for t in tuple02:
    print(t)

print(tuple02[1:3])#slicing

tuple03 = (30,27,23,45,56,78)
reversed_items = tuple03[::-1]#reversing values
print(tuple03)
print(reversed_items)

print(tuple03 * 2)
print(tuple03)
```

```python
my_list = [10, 20, 30, 40, 50]
print("List", my_list)
my_tuple = tuple(my_list)

print("Tuple",my_tuple)
print(type(my_tuple))
```

```python
fruits = ("apple", "banana", "cherry", "date")

print("cherry" in fruits)
print("mango" in fruits)
print(len(fruits))
print(fruits.count("date"))
```

```python
person = ("Alice", 30, "Engineer", "Pune")

name, age, job, city = person #unpacking Tuple values

print("Name:", name)
print("Age:", age)
print("Job:", job)
print("City:", city)

def ShowValues():
    person = ("Alice", 30, "Engineer", "Pune")
    return person

name, age, job, city = ShowValues()
print("Name:", name)
print("Age:", age)
print("Job:", job)
print("City:", city)
```

```python
scores = (88, 95, 70, 62, 99, 74, 85)

total = sum(scores)
highest = max(scores)
lowest = min(scores)

print("Sum:", total)
print("Max:", highest)
print("Min:", lowest)
```

```python
students = (("Alice", 88), ("Bob", 73), ("Charlie", 95), ("Diana", 61))

sorted_students = tuple(sorted(students, key=lambda x: x[1]))
print("Sorted:", sorted_students)
```

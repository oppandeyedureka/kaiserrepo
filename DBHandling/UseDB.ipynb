```python
# Install once in the notebook environment if needed:
# %pip install mysql-connector-python

import mysql.connector
from mysql.connector import Error

connection = None
cursor = None

try:
    connection = mysql.connector.connect(
        host="localhost", port=3306, user="root", password="Password@1234", 
        database="companydb", ssl_disabled=True, use_pure=True
    )

    if connection.is_connected():
        insert_sql = """
            INSERT INTO dept (deptid, deptname, deptloc)
            VALUES (%s, %s, %s)
        """
        values = (103, "SW Dev", "Pune")

        cursor = connection.cursor()
        cursor.execute(insert_sql, values)
        connection.commit()

        print(f"Record inserted successfully. New row ID: {cursor.lastrowid}")

except Error as error:
    if connection is not None and connection.is_connected():
        connection.rollback()
    print(f"Database error: {error}")

finally:
    if cursor is not None:
        cursor.close()
    if connection is not None and connection.is_connected():
        connection.close()
        print("MySQL connection closed.")
```

```python
import mysql.connector

try:
    connection = mysql.connector.connect(
            host='localhost', user='root', password='Password@1234',
            database='bankingdb', ssl_disabled=True, use_pure=True
    )
    if connection.is_connected():
        print("Connected to the database")
        cursor = connection.cursor()
        query = "SELECT customerid, count(*) as totalaccts, sum(balance) as custbalance FROM bankingdb.accounts GROUP BY customerid"
        cursor.execute(query)
        rows = cursor.fetchall() #select queries
        for row in rows:
            print(row)
            #print(row[0], row[1], row[2], row[3])
    
except mysql.connector.Error as err:
        print(f"Error: {err}")
```

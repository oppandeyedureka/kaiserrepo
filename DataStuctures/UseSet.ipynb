Implementing Set Methods

```python
setobj = set()#Empty set
setobj2 = {101, "Ganesh", 234234, "Pune"}
print(setobj2)

#update and modifications
#setobj2[0] = 102 #Item assignment not supported
setobj2.add(2423123)
print(setobj2)
#setobj2.insert(1, 34234)#no insert attribute
#setobj2.remove("Mumbai")#Error since Mumbai is not a value
print(setobj2)
setobj2.discard("Mumbai")
print(setobj2)
```

```python
set1 = {1,2,3,4,5,6}
set2 = {2,4,6,9,12,15}

print("Uncommon values", set1.difference(set2))
print("Uncommon values", set1 - set2)

print("Intersection/common values", set1.intersection(set2))
print("Intersection: ", set1 & set2)

print(set1.issubset(set2))
print(set1.issuperset(set2))

print("union", set1.union(set2))
print("Union : ", set1 | set2)

print("Symmetric diff",set1.symmetric_difference(set2))
print("Symmetric diff",set1 ^ set2)#symmetric difference


print(set1.pop())#remove value at zero index
print("Disjoint/Are there common values", set1.isdisjoint(set2))
print(set1)
print("Length:",len(set1))
set1.clear()#remove all values
```

```python
#List to Set conversion/Type casting
list_values = [1,5,2,8,4,3,5,6,9,6,7,8,9]
print(list_values)
numset = set(list_values)
print(numset)
print(sum(numset))
print(min(numset))
print(max(numset))
print(sorted(numset))
print(sorted(numset,reverse=True))

for s in numset:
    print(s.__hash__)

tobj = tuple(numset)
print(tobj)
```

```python
newset = {10,1,2,33,45,67,88,90,0}
print(newset)
for n in newset:
    print(n," ",n.__hash__)
newset.pop()
print(newset)
```

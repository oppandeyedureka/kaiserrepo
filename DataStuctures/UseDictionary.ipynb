Dictionary examples

```python
setObj = set()#Empty Set
dObj = {}#Empty Dictionary

dObj = {"DeptId":10,"DeptName":"Sales","DeptLoc":"Pune"}
print(dObj)
#Data Access using Keys
print("Dept Id :",dObj["DeptId"])
print("Dept Name :",dObj["DeptName"])
print("Dept Loc :",dObj["DeptLoc"])

#Update values in existing objects
dObj["DeptId"]=20
dObj["DeptLoc"]="Mumbai"
print(dObj)

print(dObj.get("DeptId"))
for key in dObj.keys():
    print("Key Name:", key, " Value: ",dObj.get(key))

Keys = dObj.keys()
print(Keys)

Values = dObj.values()
print(Values)

print(dObj.items())
for item in dObj.items():
    print(item)
```

```python
dObj2 = {
         1001:{"EmpName":"Ganesh","Location":"Pune"},
         1002:{"EmpName":"Mahesh","Location":"Mumbai"},
         1003:{"EmpName":"Dinesh","Location":"Chennai"}
        }

#print(dObj2)
print(dObj2[1001])
print(len(dObj2))
print(min(dObj2))
print(max(dObj2))
print(sum(dObj2))

print(len(dObj2[1001]))
```

```python
#Integration of types
dObj3 = {
    1001:["Hello","hi"],
    1002:("Welcome","Bye"),
    1003:{12,23,45,56}
}

for item in dObj3.items():
    print(item)

dObj4 = dObj3.copy()
dObj3.clear()

print(dObj4)

dObj4.popitem()
print(dObj4)
```

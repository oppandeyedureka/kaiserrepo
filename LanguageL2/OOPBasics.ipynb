Understand key features of OOP (Object Oriented Programming)

```python
#define class
class Employee:
    #default Constructor 
    """
    def __init__(self):
        self.empid = 1001
        self.empname = "ABC"
        self.salary = 234234.454
    """

    #Parameterized Constructor
    
    def __init__(self, empid, empname, salary):
        self.empid = empid
        self.empname = empname
        self.salary = salary
        
    #Behaviour or methods
    def showEmpDetails(self):
        print("Emp ID: ",self.empid, " Emp Name: ", self.empname, " Emp Salary: ", self.salary)
    
```

```python
#create objects of Employee
#empobj1 = Employee()
#empobj1.showEmpDetails()

empobj2 = Employee(1002, "XYZ", 123456.789)
empobj2.showEmpDetails()
```

```python
class Manager(Employee):
    def __init__(self, empid, empname, salary, perks):
        super().__init__(empid, empname, salary)
        self.perks = perks

    def showtotalsalary(self):
        total_salary = self.salary + self.perks
        print("Total Salary: ", total_salary)
```

```python
mgrobj = Manager(1003, "PQR", 200000.00, 50000.00)
mgrobj.showEmpDetails()
mgrobj.showtotalsalary()
```

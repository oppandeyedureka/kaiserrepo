```python
from sklearn import datasets
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score, classification_report, precision_score

diabetes = datasets.load_diabetes()
X = diabetes.data  # Features
y = (diabetes.target > 140).astype(int)  # Binary target: 1 if diabetes, 0 otherwise

# Create a DataFrame for better visualization
diabetes_df = pd.DataFrame(data=X, columns=diabetes.feature_names)
diabetes_df['diabetes'] = y
diabetes_df.head()

#Train Test Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

#Creating Model Class
# model = KNeighborsClassifier(n_neighbors=5) #Category based data #71
model = DecisionTreeClassifier(max_depth=15, max_leaf_nodes=10, max_features=4) #75
# model = LogisticRegression() #80

#Train the model
model.fit(X_train, y_train)

predictions = model.predict(X_test)

#Check the Accuracy/Metrics  
f1_score(y_test, predictions) #for classifier
print(classification_report(y_test, predictions,zero_division=1))#for classifier

print(precision_score(y_test, predictions, average=None, zero_division=1))
```

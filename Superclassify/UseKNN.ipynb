```python
#Import relevant Libraries and resources
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.neighbors import KNeighborsClassifier, KNeighborsRegressor
from sklearn.metrics import f1_score, root_mean_squared_error, r2_score, classification_report, precision_score

#Load Data
df = pd.read_csv("Credit.csv")
# df = pd.read_csv("house_prices.csv")

#Data Cleansing, Encoders, Scaler
# ohe = OneHotEncoder()
# location_encoded = ohe.fit_transform(df[['city']]).toarray()
# location_df = pd.DataFrame(location_encoded, columns=ohe.get_feature_names_out(['city']))

#Select Features
# X = pd.concat([df[['sqft_living', 'bedrooms', 'yr_built']], location_df], axis=1)
# y = df['price']

#For Credit.csv
X = df.drop("Credit", axis=1)
y = df["Credit"]

#Train Test Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

#Creating Model Class
# model = KNeighborsRegressor(n_neighbors=10) #Continuous values
model = KNeighborsClassifier(n_neighbors=15) #Category based data for Credit.csv

#Train the model
model.fit(X_train, y_train)

predictions = model.predict(X_test)

#Check the Accuracy/Metrics  
f1_score(y_test, predictions) #for classifier
print(classification_report(y_test, predictions,zero_division=1))#for classifier

# r2_score(y_test, predictions)
print(precision_score(y_test, predictions, average=None, zero_division=1))

new_customer = np.array([[0,25,0.202982129,91200,1,1,2,2,0,0]])
predicted_credit = model.predict(new_customer)
print(f'The predicted credit of the customer is: {predicted_credit[0]:,.2f}')

#predict the price of a house with 2000 sqft, 3 bedrooms, built in 1990, and located in 'Redmond'
# location_features = [1 if col == 'city_Redmond' else 0 for col in location_df.columns]
# new_house = np.array([[2000, 3, 1990] + location_features])
# predicted_price = model.predict(new_house)
# print(f'The predicted price of the house is: ${predicted_price[0]:,.2f}')
```

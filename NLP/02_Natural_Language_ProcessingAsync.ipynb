### Text Preprocessing

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
import re

# Download necessary resources
nltk.download('stopwords')
nltk.download('wordnet')

# Sample text
text = "Text preprocessing is an important step in Natural Language Processing. It includes removing stopwords, stemming, and lemmatization!"

# Step 1: Lowercasing
text = text.lower()
print("Lowercased Text:", text)

# Step 2: Removing Punctuation and Special Characters
text = re.sub(r'[^a-z\s]', '', text)
print("Cleaned Text:", text)

# Step 3: Removing Stopwords
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in text.split() if word not in stop_words]
print("Without Stopwords:", filtered_words)

# Step 4: Stemming
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_words]
print("Stemmed Words:", stemmed_words)

# Step 5: Lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]
print("Lemmatized Words:", lemmatized_words)
```

1. **Download resources**: We use `nltk` for text preprocessing. Download stopwords and lemmatizer resources.
2. **Lowercasing**: Convert all text to lowercase to ensure uniformity.
3. **Removing punctuation**: Use regular expressions (`re`) to remove all non-alphabetic characters.
4. **Removing stopwords**: Filter out commonly used words like "is" or "the" using NLTK's stopwords list.
5. **Stemming**: Use `PorterStemmer` to reduce words to their root form.
6. **Lemmatization**: Apply `WordNetLemmatizer` to obtain linguistically correct base forms.



### Text Tokenization


```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize

# Optional Hugging Face tokenizer
try:
    from transformers import AutoTokenizer
    hf_available = True
except ImportError:
    hf_available = False
    print("Warning: transformers is not installed. Subword tokenization will be skipped.")

# Ensure punkt is available for tokenization
try:
    nltk.data.find('tokenizers/punkt')
except LookupError:
    nltk.download('punkt')

# Sample text
text = "Natural Language Processing is fascinating. It involves text tokenization and word embeddings."
# Sentence Tokenization
sentences = sent_tokenize(text)
print("Sentence Tokens:", sentences)

# Word Tokenization
words = word_tokenize(text)
print("Word Tokens:", words)

# Subword Tokenization using Hugging Face Tokenizer
if hf_available:
    tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
    subword_tokens = tokenizer.tokenize(text)
    print("Subword Tokens:", subword_tokens)
else:
    print("Subword tokenization skipped because Transformers is not installed.")
```

1. **Sentence tokenization**: Use NLTK’s `sent_tokenize` to split text into sentences.
2. **Word tokenization**: Apply `word_tokenize` to split sentences into words.
3. **Subword tokenization**: Use Hugging Face’s BERT tokenizer to demonstrate subword splits for handling rare words and context.

```python
from gensim.models import Word2Vec
import numpy as np

# Sample sentences for training Word2Vec
sentences = [
    ["natural", "language", "processing", "is", "amazing"],
    ["word", "embeddings", "capture", "semantic", "meaning"],
    ["machine", "learning", "and", "NLP", "are", "interrelated"]
]

# Train Word2Vec Model
model = Word2Vec(sentences, vector_size=100, window=5, min_count=1, workers=2)
print("Vocabulary:", model.wv.key_to_index)

# Get Word Vector
word = "processing"
vector = model.wv[word]
print(f"Vector for '{word}':", vector)

# Similar Words
similar_words = model.wv.most_similar("language", topn=3)
print("Words similar to 'language':", similar_words)
```


1. **Training Word2Vec**: We train a Word2Vec model using sample sentences with `vector_size` (embedding size) of 100.
2. **Vocabulary**: View the list of words included in the model's vocabulary.
3. **Word Vector**: Extract the numerical vector representation of a word (e.g., "processing").
4. **Find Similar Words**: Use the model to retrieve words semantically similar to a given word.

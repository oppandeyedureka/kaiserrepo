```python
number1 = -1245
print(abs(number1))

number_value = input("Enter a number: ")#Accepts user input as a string

#number2 = abs(number_value) #Gives an error because number_value is a string, not an integer. You need to convert it to an integer before taking the absolute value.

number2 = int(number_value) #convert the string input to an integer

number2 = abs(number2) #get absolute value of the integer

#number2 = abs(int(input("Enter a number: "))) #combined single line to accept input, convert to integer, and take absolute value
print(number2)
```

```python
input_salary = input("Enter your salary: ") #Accepts user input as a string
salary = float(input_salary) #convert the string input to a float
salary = abs(salary) #get absolute value of the float
print(salary)

salary = round(salary, 2) #round the salary to 2 decimal places
print(salary)
```

```python
num1 = 234
num2= 456
print(round(num2/num1, 2)) #use round() to round the result of the division to 2 decimal places
```

```python
#min, max, len, pow
num1 = 234
num2= 456
num3 = 789
print(min(num1, num2,num3)) #use min() to get the minimum of the two
print(max(num1, num2,num3)) #use max() to get the maximum of the two

num4 = 3
print(pow(num1, num4)) #use pow() to get the power of a number. In this case, 3 raised to the power of 3

Message = input("Enter a message: ") #Accepts user input as a string
print(len(Message)) #use len() to get the length of the string

print(chr(97)) #use chr() to get the character corresponding to the ASCII value 97
```

```python
#import builtins

boolean_value = input("Enter a boolean value (True/False): ") #Accepts user input as a string
#boolvar = bool(boolean_value)
print(bool(boolean_value)) #use bool() to convert the string input to a boolean value. Note that any non-empty string will be converted to True, and an empty string will be converted to False.
```

```python
#Common Methods
str1 = "Welcome"
print(len(str1))
print("Min ASCII", min(str1))
print("Max ASCII", max(str1))
print(sorted(str1))
#A - 65
#a - 97
```

```python
#String Case Methods
str1 = "welcome to Kaiser"

print("Capitalize: ", str1.capitalize())
print("Case Fold : ", str1.casefold())

if(str1.islower()):
    print("Is Lower Case")
else:
    print("Is Mixed Case")

if(str1.isupper()):
    print("Is Upper Case")
else:
    print("Is Mixed Case")

print("Lower:", str1.lower())
print("Upper:", str1.upper())
print("Swapcase : ", str1.swapcase())
print("Title : ", str1.title())

if str1.istitle():
    print("Title Case")
else:
    print("Not Title Case")
```

```python
#String validation methods
str1 = "   W@lc0m@ t0 $tr!n?$   "
str2 = "Welc0me"
str3 = "Hello"

if str2.isalnum():
    print("Alphanumeric value")
else:
    print("No Alphanumeric values")

if str3.isalpha():
    print("Alpha value")
else:
    print("No Alpha values")

num1 = "34578"
if num1.isdecimal():
    print("Decimal value")
else:
    print("Not a decimal")

num1 = "345.78"
if num1.isdigit():
    print("Digit value")
else:
    print("Not a digit")

num1 = "345"
if num1.isnumeric():
    print("Numeric value")
else:
    print("Not a Number")

print("Check Space:",str1.isspace())
print("Printable:",str1.isprintable())
```

```python
#String methods to remove spaces and unwanted chars
str1 = "            Welcome               "
print(len(str1))
print(len(str1.lstrip()))
print(len(str1.rstrip()))
print(len(str1.strip()))

str2 = "ACC_12345"
print(str2)
print(str2.removeprefix("ACC_"))
str3 = "12345_KP.txt"
print(str3.removesuffix(".txt"))
```

```python
#Using Substring methods
emailaddress = input("Enter valid Email adress:")
print(emailaddress)

if len(emailaddress.strip())<7:
    print("Invalid email length")
else:
    #Check if @ appears only once
    if emailaddress.count("@")>1:
        print("@ is more than once")
    else:
        if(not (emailaddress.endswith(".com"))):
            print(".com not found")#find .com in email
        else:
            print("Valid Email Address")
```

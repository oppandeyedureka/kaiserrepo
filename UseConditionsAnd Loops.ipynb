Applications for using Condiitons and Loops

```python
num1 = int(input("Enter a number : "))
primeflag=True
if num1>0:
    print("+ve Number")
    if num1%2==0:
        print("The number is even number")
    else:
        for i in range(2,num1):
            if num1%i==0:
                print("Number is not prime")
                primeflag=False 
                break
            else:
                continue
        if primeflag==True:
            print("the number is prime number")
        else:
            print("The number is not prime number")
elif num1<0:
    print("the number is -ve")
else:
    print("the number is zero")
```

```python
#check if number is prime number
num1 = int(input("Enter a number:"))
primeflag = True

for i in range(2,num1):
    if num1%i==0:
        print("Number is not prime")
        primeflag=False 
        break
    else:
        continue

if primeflag==True:
    print("the number is prime number")
else:
    print("The number is not prime number")
```

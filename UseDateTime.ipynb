Notebook for Learning DateTime Type

```python
from datetime import datetime, timedelta
datevar = datetime.now()
print(datevar)

print(datetime.today())
print(datetime(2026, 7, 31))

print("Year : ", datevar.year)
print("Month : ", datevar.month)
print("Day : ", datevar.day)
print("Week day : ", datevar.weekday())

datevar2 = datetime(2026, 7, 31)

#Difference of dates
print(datevar2 - datevar)

#Date add
print(datevar + timedelta(weeks=3))#hours, days, weeks

#Convert from string to date object
datevar3 = datetime.strptime("15-8-2026","%d-%m-%Y")
print(datevar3 - datevar)

#Convert from date object to string
print("Independence day "+datetime.strftime(datevar3,"%d-%m-%Y"))
print("Week day : ", datevar3.weekday())
```

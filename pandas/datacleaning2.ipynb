```python
import pandas as pd

df = pd.read_csv("dirty_data_ship.csv")

#Describe Dataset
df.head()
df.info()
df.shape
df.columns

df = df.dropna(how='all')#remove all blank rows
#Rename Columns
df.rename(
    columns={"Order ID":"OrderID"},
    inplace=True
)

#unpivot the dataset
df_long = pd.melt(
    df,
    id_vars=['OrderID'],
    var_name='Category',
    value_name='Sales'
)

#remove empty sales records
df_long = df_long.dropna(
    subset=['Sales']
)

#remove total columns
df_long = df_long[
    ~df_long['Category']
      .str.contains('total')
]

#Split segment and Ship mode
df_long[['Segment','ShipMode']] = (
    df_long['Category']
    .str.split('_', expand=True, n=1)
)

#Standardize values
df_long['Segment'] = (
    df_long['Segment']
    .str.title()
)

df_long['ShipMode'] = (
    df_long['ShipMode']
    .str.replace('_',' ')
    .str.title()
)

#rename shipmode
df_long.rename(
    columns={
        "ShipMode":"Ship Mode"
    },
    inplace=True
)

#reorder columns
df_long = df_long[
[
'Segment',
'Ship Mode',
'OrderID',
'Sales'
]
]

#process grand total
grand_total = (
    df[df['OrderID']=="Grand Total"]
)


df_long = df_long.sort_values(
[
'Segment',
'Ship Mode',
'OrderID'
]
)

df_long
```

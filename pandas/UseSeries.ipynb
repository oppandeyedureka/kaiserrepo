Access Data from Panda Series

```python
import pandas as pd
import numpy as np

data = np.array(['a', 'b', 'c', 'd'])
s = pd.Series(data)

print(s[1])

# Display the Original series
print('Original Series:',s, sep='\n')

# Slice the range of values
result = s[1:3]

# Display the output
print('Values after slicing the Series:', result, sep='\n')

result = s[0:4:2]
print(result)
```

```python
import pandas as pd
s = pd.Series([1,2,3,4,5],index = ['a','b','c','d','e'])

#retrieve the first three element
print(s[:5]) #print(s[0:5])
```

```python
import pandas as pd
s = pd.Series([1,2,3,4,5],index = ['a','b','c','d','e'])

#retrieve the last three element
print(s[-3:])
```

```python
import pandas as pd
s = pd.Series([1,2,3,4,5],index = ['a','b','c','d','e'])

# Slice multiple elements
print(s['a':'d'])
```

Data modification in Panda series

```python
import pandas as pd
s = pd.Series([1,2,3,4,5], dtype=pd.Int64Dtype)

# Display the original series 
print("Original Series:\n",s)

# Modify the values of first two elements
s[:2] = [100, 200]

print("Series after modifying the first two elements:\n",s)
```

```python
#Panda Series Attributes
series = pd.Series([1,2,3,4,5])

print("Empty: ", series.empty)
print("Axes :", series.axes)
print("first Value: ",series.first)
#print("List conversion :", series.list(0:3))
print(series.flags)
print("Number of values :",series.size)
print("Dimensions :",series.ndim)
print("Values :\n",series.values)
print("Keys :\n", series.keys)
print("shape :",series.shape)
print("Index value :",series.index)
```

Arithmetic Operators with Series

```python
series  = pd.Series([1,2,3,4,5])
print(series*2)
print(series+2)
print(series/2)
print(series-2)
print(series**2)
```

```python
import pandas as pd
s1 = pd.Series([1,2,3,4,5],index = ['a','b','c','d','e'])
s2 = pd.Series([9, 8, 6, 5], index=['x','a','b','c'])

# Apply all Arithmetic Operations and Display the Results
print('\nAddition:\n',s1+s2)
print('\nSubtraction:\n', s1-s2)
print('\nMultiplication:\n', s1 * s2)
print('\nDivision:\n', s1/s2)
```

```python
player_points_2 = pd.Series(
    [30, 25, 15, 10, 20],  # Points scored by each player in a match
    index=["Player 1", "Player 2", "Player 3", "Player 4", "Player 5"]  # Player Names
)
print(player_points_2)

```

```python
data_dictionary = {'Player 1': 30, 'Player 2': 25, 'Player 3': 15, 'Player 4': 10, 'Player 5': 20}
player_points_3 = pd.Series(data_dictionary)

print("Head",player_points_3.head())
print("Tail",player_points_3.tail())

print(player_points_3.is_unique)
```

```python
#Descriptive statistics
print("Mean :", player_points_3.mean())
print("Median :", player_points_3.median())
print("Mode :", player_points_3.mode())
print("Std Dev :", round(player_points_3.std(),3))
print("Sum :", player_points_3.sum())
print("Min :", player_points_3.min())
print("Max :", player_points_3.max())
```

```python
print(player_points_3.sort_index())
print(player_points_3.sort_values())
```

```python
#Comparision operators
player_points_3 > 20
```

```python
null_series = pd.Series([25, 10, 5, np.nan, 8, 41, np.nan])
print(null_series.isna())

null_series_removed = null_series.dropna() #delete null values
print(null_series_removed)

null_series_filled = null_series.fillna(0) #replace null values by 0
print(null_series_filled)

null_series_filled_2 = null_series.fillna(null_series.median()) #replace null values by Median of series
print(null_series_filled_2)
```

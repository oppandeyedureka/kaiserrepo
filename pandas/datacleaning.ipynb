```python
#Data Cleaning
import pandas as pd
import numpy as np

df = pd.read_csv("my_file.csv")
print(df.columns)
df['Actual gross'] = df['Actual gross'].str.replace("$","")
df['Actual gross'] = df['Actual gross'].str.replace(",","")
df['Actual gross'] = df['Actual gross'].str.replace("[b]","")
df['Actual gross'] = df['Actual gross'].str.replace("[e]","")

df['Average gross'] = df['Actual gross'].str.replace("$","")
df['Average gross'] = df['Actual gross'].str.replace(",","")

df['Actual gross(in 2022 dollars)'] = df['Actual gross'].str.replace("$","")
df['Actual gross(in 2022 dollars)'] = df['Actual gross'].str.replace(",","")

df['Year(s)'] = df['Year(s)'].str.split("–").str[0]

#Convert cleaned String values to numeric values 
df['Actual gross'] = pd.to_numeric(df['Actual gross'],errors='coerce')
df['Actual gross(in 2022 dollars)'] = pd.to_numeric(df['Actual gross(in 2022 dollars)'],errors='coerce')
df['Average gross'] = pd.to_numeric(df['Average gross'],errors='coerce')

df.groupby("Artist").aggregate({'Actual gross':"sum"})

print(df['Actual gross'].skew())
print(df['Actual gross'].kurtosis())
print("Mean ", round(df['Actual gross'].mean(),3)) 
print("Median ", round(df['Actual gross'].median(),3))
print("Mode ", round(df['Actual gross'].mode(),3))
print("Min ", min(df['Actual gross']))
print("Max ", max(df['Actual gross']))

df.to_csv("cleanfile.csv")
```

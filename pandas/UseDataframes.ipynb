```python
import pandas as pd
productsdf = pd.read_csv("products-1000.csv")
productsdf.describe()
productsdf.shape
print(productsdf.columns)

#productsdf.groupby("Color").count()
#productsdf.groupby("Category").count()
#productsdf.groupby("Availability").count()
productsdf.groupby("Size").count()

eankurtosis = productsdf["EAN"].kurtosis()
print("Kurtosis on EAN ",eankurtosis)

print(round(productsdf["Stock"].kurtosis(),3))
print(round(productsdf["Price"].skew(),3))
print("Mean ", round(productsdf["Price"].mean(),3))
print("Median ", round(productsdf["Price"].median(),3))
print("Mode ", round(productsdf["Price"].mode(),3))
print("Min ", min(productsdf["Price"]))
print("Max ", max(productsdf["Price"]))

#dfagg = productsdf.groupby("Availability").agg({"Price":"mean","Stock":"sum"})
#dfagg = productsdf.groupby("Color").agg({"Price":"mean","Stock":"sum"})
dfagg = productsdf.groupby("Category").aggregate({"Price":"median","Stock":"sum"})

Q1 = productsdf["Stock"].quantile(0.25)
Q3 = productsdf["Stock"].quantile(0.75)

IQR = Q3-Q1
lower = Q1 - 0.5 * IQR
upper = Q3 + 0.5 * IQR

outliers = productsdf["Stock"][(productsdf["Stock"] < lower) | (productsdf["Stock"] > upper)]
print("Outliers : ",outliers.count())
```

```python
import pandas as pd

# create a sample DataFrame
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David'],
        'Age': [25, 30, 35, 40],
        'City': ['New York', 'London', 'Paris', 'Tokyo'],
        'Height': ['165', '178', '185', '171'],
        'Profession': ['Engineer', 'Entrepreneur', 'Unemployed', 'Actor'],
        'Marital Status': ['Single', 'Married', 'Divorced', 'Engaged']}
df = pd.DataFrame(data)

# display the original DataFrame
print("Original DataFrame:")
print(df)
print()

#drop / delete row
df.drop(index=3)

# delete age column
df.drop('Age', axis=1, inplace=True)

# delete marital status column
df.drop(columns='Marital Status', inplace=True)

# delete height and profession columns
df.drop(['Height', 'Profession'], axis=1, inplace=True)

# display the modified DataFrame after deleting rows
print("Modified DataFrame:")
print(df)
```

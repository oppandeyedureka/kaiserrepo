Working with various Window functions

```python
#Rolling Window
import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.rand(10,4), 
index=pd.date_range('1/1/2026', periods=10),
columns=['A','B','C','D'])

dford = pd.read_excel("Online-Store-Orders.xlsx")
#dford.describe()
#dford.head()
dford.shape

#print(df)
print(dford["TotalPrice"].rolling(window=5).mean())
```

```python
import pandas as pd
weather = pd.DataFrame({'Temperature':[57, 54, 60, 54, 57, 58, 52, 52, 59, 54, 53, 57, 56, 60, 55, 58, 59, 64, 65, 66, 67, 74]})

#print("Temp Mean :", round(weather.rolling(window=3).mean()),2)
#print("Temp Median :", weather.rolling(window=3).median())
#print("Temp Sum :", weather.rolling(window=3).sum())
print("Temp Min :", weather.rolling(window=3).min())
```

```python
#Use Weighted Window

import pandas as pd
#dataset = pd.Series(range(100))
#result = dataset.rolling(window=3, win_type="triang").mean()#gaussian, bartlett, triang
#print(result)

dford = pd.read_excel("Online-Store-Orders.xlsx")
result = dford["TotalPrice"].rolling(window=5, win_type="triang").mean()
print(result)
```

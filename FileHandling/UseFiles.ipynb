```python
#fileref = open("File01.txt","w")#Open file in write Mode
fileref = open("File01.txt","a+")#Open file in append mode
fileref.write("Welcome to File Handling")
strings = ["Hello", "Hi", "Welcome"]
fileref.writelines(strings)
fileref.close()
```

```python
filerefread = open("File01.txt","r")#open file in read mode
print(filerefread.read())
for line in filerefread.readlines():
    print(line)
filerefread.close()
```

```python
#Automatically close file after Reading/Writing
with open('File01.txt', 'r') as file:
    content = file.read()
print(content)
```

```python
# Step 1
import pickle
 
config_dictionary = {'remote_hostname': 'google.com', 'remote_port': 80}
 
# Step 2
with open('config.dictionary', 'wb') as config_dictionary_file:
 
  # Step 3
  pickle.dump(config_dictionary, config_dictionary_file)
```

```python
# Step 1
import pickle
 
# Step 2
with open('config.dictionary', 'rb') as config_dictionary_file:
 
    # Step 3
    config_dictionary = pickle.load(config_dictionary_file)
 
    # After config_dictionary is read from file
    print(config_dictionary)
```

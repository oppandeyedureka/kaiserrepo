```python
value = int(input("Enter value between 1-3:"))
match value:
    case 1:
        print("Value entered is 1")
    case 2:
        print("Value entered is 2")
    case _:
        print("Value is not taken care")    
```

```python
UserInput = input("Enter the operation to perform:")
match UserInput:
    case "Add Doctor":
        print("Doctor Added")
        doctype = input("Enter doctor type")
        match doctype:
            case "OPD":
                print("OPD Doctor")
            case "IPD":
                print("IPD doctor")
    case "Add Nurse":
        print("Nurse Added")
    case _:
        print("Value not expected")
```

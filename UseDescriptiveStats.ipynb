```python
from math import sqrt
#variance calculation
sum = 360000 + 160000 + 2560000 + 1960000 +160000
count = 5
variance = sum/(count-1)
print(variance)

stddev = sqrt(variance)
print(round(stddev,3))
```

```python
#calculate Variance
data = [5, 7, 9, 10, 14, 15]

sum=0
for i in data:
    sum += i
MeanValue = sum / len(data)
print(MeanValue)

#variance
sumsqrs = 0
for i in data:
    diffval = i - MeanValue
    sumsqrs += pow(diffval,2)
variance = sumsqrs/len(data)
print(round(variance,2))
```

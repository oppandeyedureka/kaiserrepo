```python
import pandas as pd
import numpy as np
from sklearn.model_selection import TimeSeriesSplit
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# Simulate real-world dataset
np.random.seed(42)
n_customers, n_months = 100, 24

customer_ids = np.repeat(np.arange(1, n_customers+1), n_months)
months = np.tile(np.arange(1, n_months+1), n_customers)
default = np.random.binomial(1, 0.2, n_customers*n_months)

feature_1 = np.random.randn(n_customers*n_months)
feature_2 = np.random.randn(n_customers*n_months)

df = pd.DataFrame({
    'customer_id': customer_ids,
    'month': months,
    'feature_1': feature_1,
    'feature_2': feature_2,
    'default': default
})
df = df.sort_values(['customer_id', 'month'])

X = df[['feature_1', 'feature_2']]
y = df['default']

tscv = TimeSeriesSplit(n_splits=5)

for i, (train_idx, test_idx) in enumerate(tscv.split(X)):
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    model = RandomForestClassifier(random_state=42)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)

    print(f"Fold {i}: Train indices={train_idx[:3]}..., Test indices={test_idx[:3]}..., Accuracy={acc:.3f}")
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import TimeSeriesSplit, cross_val_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
from sklearn.tree import DecisionTreeRegressor

df = pd.read_csv('monthly-milk-production-pounds.csv',parse_dates=['Month'], index_col='Month')
X = df.sort_values('Month')
df1 = pd.read_csv('monthly-milk-production-pounds.csv')
Y = df1['Production']

CV = TimeSeriesSplit(n_splits=10, max_train_size=100, test_size=5, gap=10)

# # model = RandomForestClassifier(random_state=42)
model = DecisionTreeRegressor(random_state=42)

score = cross_val_score(model, X, Y, cv=CV, error_score='raise', scoring='accuracy')
score_mean = score.mean()
print(f"Cross-Validation Score: {score_mean:.4f}")
```

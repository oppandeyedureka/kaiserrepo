```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.datasets._california_housing import fetch_california_housing
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
from sklearn.model_selection import train_test_split, RandomizedSearchCV
from sklearn.metrics import classification_report, r2_score
from sklearn.model_selection import cross_val_score
from sklearn.svm import SVC, SVR
import optuna
```

```python
def objective(trial):
    housing = fetch_california_housing()
    cancer = load_breast_cancer()
    # svc_c = trial.suggest_loguniform('C', 1e-3, 1e3)
    svc_c = trial.suggest_float('C', 1e-3, 1e3, log=True)

    # svc_gamma = trial.suggest_loguniform('gamma', 1e-4, 1e-1, log=True)
    svc_gamma = trial.suggest_float('gamma', 1e-4, 1e-1, log=True)

    # clf = SVC(C=svc_c, gamma=svc_gamma)
    clf = SVR(C=svc_c, gamma=svc_gamma)

    # score = cross_val_score(clf, cancer.data, cancer.target, n_jobs=-1, cv=3)
    score = cross_val_score(clf, housing.data, housing.target, n_jobs=-1, cv=3)
    accuracy = score.mean()
    return accuracy
```

```python
study = optuna.create_study(direction='maximize')
study.optimize(objective, n_trials=10)

print(f"Best parameters: {study.best_params}")
print(f"Best accuracy: {study.best_value}")
```
